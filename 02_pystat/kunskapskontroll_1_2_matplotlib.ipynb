### Explain shortly what a figure, axes, axis and an artist is in Matplotlib.

```python
# Figure: Is the whole figure that we can see it as a canvas. 
# Axes: Is the part of the canvas that we have our plot attached on it. We can do the plotting directly from the Axes wich give us flexibility to customize plots.
# Axis: Is the numerical scale like(x-axis, y-axis), that runs along the edges of an "Axes". 
# Artist: Everything we see on the figure is an "Artist", like figure, axes and axis. For the most of the time the "Artist" are attached to an "Axess"
```

### Create a plot of the function y = x^2 [from -4 to 4, hint use the np.linspace function] both in the object-oriented approach and the pyplot approach. Your plot should have a title and axis-labels.

```python
# The OO-approach
x = np.linspace(-4, 4, 200)
y = x**2

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("y = x^2")
ax.set_xlabel("x")
ax.set_ylabel("y")
```

```python
# the pyplot approach
plt.plot(x, y)
plt.title("y = x^2")
plt.xlabel("x")
plt.ylabel("y")
```

```python
np.random.seed(15)
random_data_x = np.random.randn(1000)
random_data_y = np.random.randn(1000)
x = np.linspace(-2, 2, 100)
y = x**2

fruit_data = {'grapes': 22, 'apple': 8, 'orange': 15, 'lemon': 20, 'lime': 25}
names = list(fruit_data.keys())
values = list(fruit_data.values())

fig, axs = plt.subplots(1,2, layout="constrained")
fig.suptitle("Two types of Plots", size = 20)

axs[0].scatter(random_data_x, random_data_y)
axs[0].set_title("Scatter Plot")

axs[1].bar(names, values)
axs[1].set_title("Bar Plot")
                        
```
